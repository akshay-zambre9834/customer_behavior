```python
##snackcase = all are in smallcase with underscore 
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
```

```python
df.columns
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'}) ## rename the column name or we say update
```

```python
## create a column of age_group 
labels = ["young adult","adult","middle aged","senior"]
df['age group']=pd.qcut(df['age'],  q=4,   labels=labels)
```
